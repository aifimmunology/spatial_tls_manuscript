## Load adata

```python
adata = sc.read_h5ad(os.path.join(out_dir,'adata_distance.h5ad'))
```

## Save final processed adata object for manuscript

```python
# Remove adata.obs columns with only nan values - these are carried over from merging with the single cell reference
nan_cols = adata.obs.columns[adata.obs.isna().all()]
print(nan_cols.tolist())
adata.obs = adata.obs.loc[:, ~adata.obs.isna().all()]
```

```python
# declutter - only keep relevant columns
cols_delete = ['log1p_n_genes_by_counts', 'log1p_total_counts',
       'pct_counts_in_top_10_genes', 'pct_counts_in_top_20_genes',
       'pct_counts_in_top_50_genes', 'pct_counts_in_top_150_genes', 
       '_scvi_batch', '_scvi_labels', 'celltype_scanvi', 'C_scANVI',
               'leiden_scvi_refalign', 'cluster_cellcharter_8',
       'cluster_cellcharter_9', 'cluster_cellcharter_10',
       'cluster_cellcharter_11', 'cluster_cellcharter_13']
print(f"deleting {cols_delete}")

adata.obs.drop(columns=cols_delete, inplace=True)
```
