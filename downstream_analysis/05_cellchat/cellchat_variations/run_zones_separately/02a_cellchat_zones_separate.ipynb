### Break adata objects up by zone

```python
for zone in adata.obs['zone_consol'].cat.categories:
    print(zone)
    adata_zone = adata[adata.obs['zone_consol']==zone, :].copy()
    print(adata_zone.shape)
    filepath = os.path.join(dist_out_dir_cellchat,f"adata_cellchat_prepped_{zone}.h5ad")
    print(f"saving {filepath}")
    adata_zone.write_h5ad(filepath)
    
```
