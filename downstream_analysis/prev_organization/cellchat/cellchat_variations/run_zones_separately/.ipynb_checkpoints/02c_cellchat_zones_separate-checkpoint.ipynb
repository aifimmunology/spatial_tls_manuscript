### Run netAnalysis_computeCentrality

```r
cellchat_tls <- netAnalysis_computeCentrality(cellchat_tls, slot.name = "netP")
cellchat_bronchi <- netAnalysis_computeCentrality(cellchat_bronchi, slot.name = "netP")
cellchat_vessels <- netAnalysis_computeCentrality(cellchat_vessels, slot.name = "netP")
cellchat_adv <- netAnalysis_computeCentrality(cellchat_adv, slot.name = "netP")
cellchat_paren <- netAnalysis_computeCentrality(cellchat_paren, slot.name = "netP")
```

## Plotting

```r
# create named list of cellchat objects
chat_list <- list(
    "TLS" = cellchat_tls,
    "bronchi" = cellchat_bronchi,
    "vessels" = cellchat_vessels,
    "adventitia" = cellchat_adv,
    "parenchyma" = cellchat_paren
)
```

```r
# First adjust the cell label order
label_df <- read.csv("/home/workspace/spatial_tls_manuscript/inputs/20250519_mouse_lung_label_mapping_ordered.csv")

## Update cell order and palette
for (obj_name in names(chat_list)) {
  
    obj <- chat_list[[obj_name]]

    # Get the current labels in the CellChat object
    cellchat_labels <- unique(obj@idents)
    print(length(cellchat_labels))
    
    # Keep only labels that are in the cellchat object
    ordered_labels <- label_df$label_fine[label_df$label_fine %in% cellchat_labels]

    # reverse 
    cell_order <- rev(ordered_labels)
    
    # Update the CellChat object with new label order
    obj@meta$labels <- factor(obj@idents, levels = cell_order)
    obj <- updateClusterLabels(obj, new.order = cell_order)

    celltype_palette_mapped_zone <- celltype_palette_mapped[cell_order]

    ###
    
    # check for sig interactions
    df.net <- NULL
    # Try to subset the communication
    df.net <- tryCatch({
    subsetCommunication(obj, pairLR.use = single_interaction)
    }, error = function(e) {
    # This code runs ONLY if an error occurs
    return(NULL) 
    })
    
    # Check if df.net is NULL or has 0 rows before plotting
    if (is.null(df.net) || nrow(df.net) == 0) {
    message(paste("Skipping", obj_name, ": No significant interactions for this pair."))
    next
    }
  
  # Generate the plot (using netVisual_aggregate as an example)
  # We use the name from the list to dynamically title the plot
    print(obj_name)
    
    # 1. Capture the Bubble Plot
    p1 <- netVisual_bubble(
        obj,
        pairLR.use = single_interaction,
        remove.isolate = TRUE, 
        title.name = paste0(obj_name)
    )
    print(p1)
    
    # Save the Bubble Plot
    ggsave(
        filename = paste0(plot_out_dir,"/", obj_name, "_bubble_plot.pdf"), 
        plot = p1, 
        width = 8, height = 6
    )

    # 2. Capture the Individual Circle Plot
    # Note: netVisual_individual returns a plot object, but also opens a device. 
    # For saving, we often use pdf/png devices directly for circle plots 
    # because they use base R graphics rather than pure ggplot2.
    
    plot_path_circle <- paste0(plot_out_dir,"/", obj_name, "_circle_plot.pdf")
    
    pdf(plot_path_circle)
    netVisual_individual(
        obj,
        signaling = "IL2",          
        pairLR.use = single_interaction$interaction_name, 
        layout = "circle",
        color.use = celltype_palette_mapped_zone
    )
    dev.off() 

    message(paste("Successfully saved plots for:", obj_name))
}
```
